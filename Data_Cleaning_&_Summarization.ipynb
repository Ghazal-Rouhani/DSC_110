Treating missinginess

1- Dropping the rows with missing values

2- Fill in with the mean

dropna() removes rows that contain at least one missing value. we are saving the result in a new dataframe called df_complete. This way, we can keep the original dataset without removing anything from it.

```python
#drop all rows that have any missing value
df_complete = df.dropna()
```

Here we are checking how many rows are left after removing incomplete rows. This helps us see how much data was removed during cleaning.

```python
df_complete.shape[0]
```

We are checking the percentage of missing values again. Since we removed all rows with missing data, every column should now show 0%.

```python
df_complete.isna().mean()*100
```

Here we only want to remove rows where BMI is missing. By using subset, we can focus on the BMI column instead of the entire dataset. Rows with missing values in other columns will still be included.

```python
#dropping only rows where select fields have missing values
df_complete_bmi = df.dropna(subset = ['bmi'])
```

here checking the BMI column again to make sure the missing values were removed. If everything worked correctly, the result should be zero.

```python
#Verify it
df_complete_bmi['bmi'].isna().sum()
```

Another way to handle missing data is to replace missing values with the mean. Here we are calculating the average systolic blood pressure. Pandas automatically ignores missing values when calculating the mean.

```python
#Filling na with mean, zero, or with any other value
#Get the mean and assign it to a variable
mean_bp = df['systolic_bp'].mean()
```

here we are printing the average systolic blood pressure. We use round() to show only two decimal places.

```python
print(round(mean_bp,2))
```

here we are using fillna() to replace missing blood pressure values with the mean. This allows us to keep rows that would otherwise have missing blood pressure data. We are saving the changes directly in the original dataframe.

```python
df['systolic_bp'] = df['systolic_bp'].fillna(mean_bp)
```

Here we check the blood pressure column again. This helps me confirm that fillna() worked and there are no missing values left.

```python
#Verify that the imputation/fillna worked
df['systolic_bp'].isna().sum()
```

Transform variables

We are creating a function that puts BMI values into different categories. The function checks the BMI number and assigns a category based on its value. It uses if, elif, and else to check the conditions in order. The function then returns the category that matches the BMI.

```python
def bmi_check(bmi):
  if bmi < 18.5:
    bmi_result = "Underweight"
  elif bmi < 25:
    bmi_result = "Normal"
  elif bmi < 30:
    bmi_result = "Overweight"
  else:
    bmi_result = "Obese"
  return bmi_result
```

here we are applying my BMI function to the BMI column. apply() runs the function for every BMI value in the dataset. The results are saved in a new column called bmi_category.

```python
df['bmi_category'] = df['bmi'].apply(bmi_check)
```

We are using value_counts() to see how many participants are in each BMI category. This helps us understand how the BMI values are distributed. We also need to remember that missing BMI values were classified as Obese by this function.

```python
#Verify it
#value_counts()
df['bmi_category'].value_counts()
```

Another way to categorize BMI is by using pd.cut(). Instead of writing a function, we can define the BMI ranges directly. bins sets the boundaries for each category. labels assigns a name to each range. This method also keeps missing BMI values as NaN instead of assigning them a category.

```python
#New function called "cut"
df['bmi_category2'] = pd.cut(df['bmi'],
                             bins = [0, 18.5, 25, 30, 100],
                             labels = ["Underweight", "Normal", "Overweight", "Obese"])
```

Here we are checking how many participants fall into each BMI category. I can compare these results with the categories created by our first function. By default, value_counts() does not include missing values.

```python
#Verify it
df['bmi_category2'].value_counts()
```

here we are including missing values in the count. dropna=False allows NaN values to appear in the results. This helps me see how many participants did not have a BMI value recorded.

```python
df['bmi_category2'].value_counts(dropna = False)
```
