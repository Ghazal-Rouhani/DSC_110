```python
# print() displays text or other information as an output.
# This is the first simple Python command and prints "Hello World!".
print('Hello World!')
```

```python
# A variable stores information that we can use later.
# Here, the variable "name" stores a text value.
name = "Ghazal"
```

```python
# This prints the value that is stored inside the variable "name".
print(name)
```

```python
# print() can display both text and the value of a variable together.
print("My name is", name)
```

```python
# This creates another variable called "age".
# Since 24 is a number and does not have quotation marks, Python treats it as a number.
age = 24
```

```python
# This prints the current value stored in "age".
print(age)
```

```python
# We can use variables in mathematical operations.
# This adds 2 to my age without changing the original value stored in "age".
print(age + 2)
```

```python
# Here I create a new variable using another variable.
# I am graduating at the same age this year, so I add 0 to my current age.
age_at_graduation = age + 0

# This displays the value stored in age_at_graduation.
print(age_at_graduation)
```

```python
# We can combine several pieces of text and variables in one print statement.
print("My age is", age, "and my name is", name, ". My age at graduation is", age_at_graduation)
```

```python
# An f-string is another way to combine text and variables.
# The "f" before the quotation marks allows variables inside {} to be replaced
# with their actual values.
print(f"My name is {name}")
```

```python
# An f-string can include multiple variables in the same sentence.
print(f"My name is {name} and my age is {age} and I'm going to graduate when I'm {age_at_graduation}.")
```

```python
# Python starts counting positions from 0.
# name[0] returns the first character of my name.
print(name[0])
```

```python
# [0:2] is called slicing.
# It starts at position 0 and stops before position 2.
print(name[0:2])
```

```python
# A negative index starts counting from the end of the string.
# -1 gives the last character of my name.
print(name[-1])
```

```python
# A list lets us store multiple values inside one variable.
# Lists use square brackets [] and values are separated by commas.
ages = [18, 19, 22, 20, 18]
```

```python
# This prints the entire list of ages.
print(ages)
```

```python
# List positions also start at 0.
# ages[0] returns the first item in the list.
print(ages[0])
```

```python
# Lists can be modified after they are created.
# This changes the first value in the list from 18 to 19.
ages[0] = 19

# Printing the list again lets us see the change.
print(ages)
```

```python
# Lists can contain text as well as numbers.
# Here I created a list using names that are familiar to me.
names = ["Ghazal", "Matt", "Mahdi", "Jiyar", "Janet"]
```

```python
# A negative index can also be used with lists.
# -1 returns the last item in the list.
print(names[-1])
```

```python
# A dictionary stores information as key-value pairs.
# Here, each person's name is a key and their age is the value.
patient_ages = {"Ghazal": 24, "Matt": 26}
```

```python
# Unlike a list, we can access a dictionary value using its key.
# Using the key "Ghazal" returns the age connected to Ghazal.
print(patient_ages["Ghazal"])
```

```python
# This variable stores a systolic blood pressure measurement.
systolic_bp = 120

# An if statement allows Python to make a decision.
# If the blood pressure is greater than 120, it prints "high".
# Otherwise, the else statement prints "normal".
if systolic_bp > 120:
    print("high")
else:
    print("normal")
```

```python
# A function is reusable code that performs a specific task.
# "def" tells Python that we are creating a function.
# This function takes a number as input and returns that number plus 1.
def add_one(number):
    return number + 1
```

```python
# Here I call the function and give it 3 as the input.
# The function should return 4.
add_one(3)
```

```python
# This function checks a systolic blood pressure value.
# It uses an if/else statement to decide whether the result is high or normal.
def bp_check(systolic_bp):

    if systolic_bp > 120:
        bp_result = "high"

    else:
        bp_result = "normal"

    # return sends the final result back when the function is called.
    return bp_result

#Question: what is this line that shows up in some of the code boxes?
```

```python
# I am testing the function using a systolic blood pressure of 170.
# Since 170 is greater than 120, the result should be "high".
bp_check(170)
```

```python
# This list contains several blood pressure measurements.
# Instead of checking each one separately, I can use a loop.
bp_list = [120, 135, 170, 110, 100]
```

```python
# A for loop repeats the same task for every item in a list.
# "bp" represents one blood pressure value from bp_list at a time.
# Each value is checked using the bp_check() function.
for bp in bp_list:
    print(bp_check(bp))
```

```python
# This function takes a BMI number as input and returns its BMI category.
# elif allows us to check several conditions in order.
def bmi_check(bmi):

    # BMI values below 18.5 are classified as underweight.
    if bmi < 18.5:
        bmi_result = "underweight"

    # If the first condition is false, Python checks if BMI is below 25.
    elif bmi < 25:
        bmi_result = "normal"

    # If BMI is not below 25, Python checks if it is below 30.
    elif bmi < 30:
        bmi_result = "overweight"

    # Any BMI that does not meet the previous conditions is classified as obese.
    else:
        bmi_result = "obese"

    # The function returns the category that was selected.
    return bmi_result
```

```python
# I am testing the BMI classification function using a BMI of 26.
# Based on the conditions above, the result should be "overweight".
bmi_check(26)
```

```python
# This function calculates BMI using weight and height.
# BMI is calculated as weight in kilograms divided by height in meters squared.
def calculate_bmi(weight, height):

    bmi = weight / (height ** 2)

    # The calculated BMI value is returned as the output.
    return bmi
```

```python
# I test the function using a weight of 60 kg and a height of 1.60 meters.
calculate_bmi(60, 1.60)
```
